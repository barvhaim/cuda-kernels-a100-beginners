# Notebook 0: Understand Indexing Before Using a GPU

Build the mental model with ordinary Python first. This notebook does not require CUDA or a GPU.


```python
n = 1000
threads_per_block = 256
blocks = (n + threads_per_block - 1) // threads_per_block
launched_threads = blocks * threads_per_block
print({"blocks": blocks, "launched_threads": launched_threads, "extra_threads": launched_threads - n})
```

Every `(blockIdx, threadIdx)` pair receives a global index. Remember: the index starts at 0, while the human-counted position is one larger.


```python
rows = []
for block_idx in range(blocks):
    for thread_idx in range(threads_per_block):
        i = block_idx * threads_per_block + thread_idx
        rows.append((block_idx, thread_idx, i, i < n))

for row in rows[:3] + rows[255:258] + rows[-3:]:
    print(row)
```

```python
assert rows[0] == (0, 0, 0, True)
assert rows[266] == (1, 10, 266, True)
assert sum(valid for _, _, _, valid in rows) == n
assert sum(not valid for _, _, _, valid in rows) == 24
print("PASS: indexing and boundary model")
```
