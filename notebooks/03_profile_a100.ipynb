# מחברת 3: profiling בסיסי עם Nsight Compute

המטרה אינה לרדוף אחרי ציון occupancy, אלא ללמוד לענות: איפה הזמן הולך ומה מגביל את ה-kernel?

```python
from pathlib import Path
import shutil
import subprocess

def find_repo_root(start: Path) -> Path:
    start = start.resolve()
    for candidate in (start, *start.parents):
        if (candidate / "CMakeLists.txt").is_file():
            return candidate
    raise RuntimeError("Run this notebook from inside the cuda-kernels-a100-beginners repository")

ROOT = find_repo_root(Path.cwd())
ncu = shutil.which("ncu")
print("ncu:", ncu)
assert ncu, "Nsight Compute CLI (ncu) is not available"
```

```python
subprocess.run([
    "cmake", "-S", str(ROOT), "-B", str(ROOT / "build"),
    "-DCMAKE_BUILD_TYPE=Release", "-DCMAKE_CUDA_ARCHITECTURES=80"
], check=True)
subprocess.run(["cmake", "--build", str(ROOT / "build"), "-j"], check=True)
```

הרצה ראשונה משתמשת בסט בסיסי ושומרת את הפלט כטקסט לניתוח. הרשאות profiling עשויות להיות מוגבלות בשרת משותף.

```python
command = [ncu, "--set", "basic", "--kernel-name", "vector_add", str(ROOT / "build/06_vector_add")]
profile = subprocess.run(command, text=True, capture_output=True)
print(profile.stdout)
print(profile.stderr)
assert profile.returncode == 0, "ncu failed; inspect permissions and stderr above"
```
